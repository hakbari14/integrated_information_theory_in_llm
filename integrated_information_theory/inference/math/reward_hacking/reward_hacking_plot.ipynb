```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('settings_1/settings_1_gsm8k.csv')
df['completion_portion_size'] = df['completion_portion_size'].astype(float)
accuracy_per_portion = df.groupby('completion_portion_size')['Accuracy'].mean()
plt.figure(figsize=(10, 6))
plt.plot(accuracy_per_portion.index, accuracy_per_portion.values, marker='o', linestyle='-', color='b')
plt.title('Accuracy vs Completion Portion Size')
plt.xlabel('Completion Portion Size')
plt.ylabel('Accuracy')
plt.grid(True)
plt.xticks(accuracy_per_portion.index)  
plt.show()
```
